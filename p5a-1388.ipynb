```python
import pandas as pd

datos = {
    "Nombre": ["Ana", "Luis", "María", "Carlos", "Sofía"],
    "Calificacion": [9, 8, 10, 7, 9]
}

# DataFrame
df = pd.DataFrame(datos)

print("=== CALIFICACIONES NC = 1388===")
# Imprimir información de datos
print(df)

print("\npromedio y maximo")
# Se usa "Calificacion" con 'C' mayúscula
print("promedio: ", df["Calificacion"].mean())
print("calificacion mayor: ", df["Calificacion"].max())
print("-+-+-+-+-+-+-Jupyter Barrera-+-+-+-+-+-+-")
print("-+-+-+-+-+-+-Jupyter 1388-+-+-+-+-+-+-")
```
